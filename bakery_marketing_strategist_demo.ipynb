LLMs are powerful to answer general knowledge, but don't have any information on your own business.

Databricks makes it easy to develop custom tools which can be called by your LLM, creating a complete agent with reasoning capability.

But that is not the end of the story. For your agent to be ready to present to your CIO, you need proper unit and regression testing. As you will see Databricks made it easy for you thanks to Databricks Monitoring and the integration with the rest of the platform with dashboard including relevant KPIs that can be easily shared.


### Build Simple UC Tools
In this notebook, we'll register the following to Unity Catalogs:
- **SQL Functions**: Create queries that access customer information, order and subscriptions.
- **Vector Search**: Create a Vector Search Indexed on your unstructured data: docs, articles, solved tickets, FAQ etc.
- **MCP**: Allow your agent to take action by connecting it to an MCP to discover and use tools such as Slack, CRM, custom APIs etc **without loosing governance**.
- **Multi-Agent supervisor**: Create an orchestrator agent that will route the request and flow of information to the different specialized agents.

## The challenge
The scenario you’ll be working with is this: you are a regional bakery franchise manager, and your goal is to win back high‑value customers who used to visit frequently but recently stopped. The agent you'll build will analyze the bakery data to find those customers and then help you draft customized emails to bring them back to your stores.

## 1/ Create your tools

Let's start by creating our first tools. We'll create a few functions to help your agent:
- compute high‑value at‑risk customers.

- get customer profile information.

- get customer purchase history.

- list available products for a given franchise, so the agent doesn't hallucinate items. 

Remember, this could be any functions!
